```python
import numpy as np
import mne

# SuS pipeline imports
import os
import sys
sys.path.append(os.path.abspath(".."))
from notebook_utils import reload_classes
from eegkit.models.subject_model import EEGSubjectModel
from eegkit.models.dtos import (
    TaskDTO, EpochPSDParamsDTO, EvokedParamsDTO, SubjectFilterDTO, EvokedJointParamsDTO
)
from IPython.display import display
```

```python
# Try to keep MNE sample evokeds as fallback, but prefer SuS via eegkit if available
import os


# Preferred: Surround Suppression (SuS) pipeline using eegkit
EEGController, EEGUI = reload_classes()
data_dir = f'/mount/NAS-public-dataset/HBN-EEG'

sm = EEGSubjectModel(data_dir=data_dir)
ctrl = EEGController(sm)
task_dto = SubjectFilterDTO(task="surroundSupp", 
                            ccd_accuracy_range=(80,None), 
                            ccd_response_time_range=(1.6,None),
                            subject_limit=1)

conds = [
    f"bg{b}_fg{f:.1f}_stim{s}"
    for b in (0, 1)
    for f in (0.0, 0.3, 0.6, 1.0)
    for s in (1, 2, 3)
]
evokeds_list = [] 
for lab in conds:
    p = EvokedParamsDTO()
    p.stimulus = lab
    evk = ctrl.get_evoked(task_dto, p)
    evokeds_list.append(evk)
```

```python

sus_labels = [lab for lab, evk in zip(conds, evokeds_list) if evk is not None]
print("SuS Conditions (preview),", sus_labels)
```

```python
# Build mapping of evoked responses using evokeds_list (SuS only)
evks = {lab: evk for lab, evk in zip(conds, evokeds_list) if evk is not None}
# Use first SuS condition throughout
key = sus_labels[0]
```

```python
# Compare all SuS evokeds with explicit colors to avoid color-cycle errors
import matplotlib.pyplot as plt

def custom_func(x):
    return x.max(axis=1)

# Build a color map large enough for many conditions
keys = list(evks.keys())
cmaps = [plt.get_cmap("tab20"), plt.get_cmap("tab20b"), plt.get_cmap("tab20c")]
palette = []
for cmap in cmaps:
    palette.extend([cmap(i) for i in range(cmap.N)])
colors = {k: palette[i % len(palette)] for i, k in enumerate(keys)}

for combine in ("mean", "median", "gfp", custom_func):
    mne.viz.plot_compare_evokeds({k: evks[k] for k in keys}, picks="eeg", combine=combine, colors=colors)
```

```python
# Compare at a single electrode with explicit colors
import matplotlib.pyplot as plt
keys = list(evks.keys())
cmaps = [plt.get_cmap("tab20"), plt.get_cmap("tab20b"), plt.get_cmap("tab20c")]
palette = []
for cmap in cmaps:
    palette.extend([cmap(i) for i in range(cmap.N)])
colors = {k: palette[i % len(palette)] for i, k in enumerate(keys)}

mne.viz.plot_compare_evokeds(
    {k: evks[k] for k in keys},
    picks="E78",
    colors=colors,
    linestyles=None,
    time_unit="ms",
)
```
